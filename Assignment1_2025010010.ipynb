c. Perform image resizing.

```python
import cv2
from matplotlib import pyplot as plt
```

```python
img = cv2.imread("IMG_1.jpg")
```

```python
resized_img = cv2.resize(img, (300, 300))
```

```python
original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
resized = cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(resized)
plt.title("Resized Image")
plt.axis("off")

plt.show()
```

```python
cv2.imwrite("resized_image.jpg", resized_img)
print("Resized image saved successfully.")
```

d. Convert a Colored Image into a Grayscale Image.

```python
import cv2
from matplotlib import pyplot as plt
```

```python
img = cv2.imread("IMG_1.jpg")
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray_img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("grayscale_image.jpg", gray_img)
print("Grayscale image saved successfully.")
```

e. Read and Display a Video

```python
import cv2
```

```python
cap = cv2.VideoCapture("vid.mp4")
```

```python
while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        print("End of video or cannot read the video.")
        break

    cv2.imshow("Video", frame)

    # Press 'q' to exit
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
```

```python
cap.release()
cv2.destroyAllWindows()
```

f. Extract Frames from a Video.

```python
import cv2
```

```python
cap = cv2.VideoCapture("vid.mp4")
```

```python
frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    # Save each frame as an image
    filename = f"frame_{frame_count}.jpg"
    cv2.imwrite(filename, frame)

    frame_count += 1

print("Frames extracted successfully.")
```

```python
cap.release()
```
