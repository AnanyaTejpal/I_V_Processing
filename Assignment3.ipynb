**1. WAP to draw a border around the input image (pad the input image).**

Input the width of the border (in terms of pixel count) and the desired colour of the border from the user. Try this for both binary and grayscale image.

In a grayscale image a colour is a single intensity, so the border colour is taken as a value from 0 (black) to 255 (white). A binary image can only hold 0 or 255, so the entered value is rounded to whichever is nearer.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("IMG.jpg", 0)

plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```

```python
width = int(input("Enter border width: "))
color = int(input("Enter border color (0-255): "))

gray_border = cv2.copyMakeBorder(image, width, width, width, width,
                                 cv2.BORDER_CONSTANT, value=color)

_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

binary_border = cv2.copyMakeBorder(binary, width, width, width, width,
                                   cv2.BORDER_CONSTANT, value=color)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_border, cmap="gray")
plt.title("Grayscale Border")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary_border, cmap="gray")
plt.title("Binary Border")
plt.axis("off")

plt.show()
```

**2. WAP for Image Compliment.**

Try this for both binary and grayscale image.

$$Complement = 255 - \text{pixel value}$$

```python
gray_complement = cv2.bitwise_not(image)
binary_complement = cv2.bitwise_not(binary)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_complement, cmap="gray")
plt.title("Grayscale Complement")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary_complement, cmap="gray")
plt.title("Binary Complement")
plt.axis("off")

plt.show()
```
