```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("IMG.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
binary = np.where(gray > gray.mean(), 255, 0).astype(np.uint8)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1); plt.imshow(gray, cmap="gray");   plt.title("Grayscale"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(binary, cmap="gray"); plt.title("Binary");    plt.axis("off")
plt.show()
```

**1. WAP to draw a border around the input image (pad the input image).**

Input the width of the border (in terms of pixel count) and the desired colour of the border from the user. Try this for both binary and grayscale image.

In a grayscale image a colour is a single intensity, so the border colour is taken as a value from 0 (black) to 255 (white). A binary image can only hold 0 or 255, so the entered value is rounded to whichever is nearer.

```python
w = int(input("Enter the border width in pixels: "))
colour = int(input("Enter the border colour (0 = black ... 255 = white): "))

colour_binary = 255 if colour >= 128 else 0
print("Border colour used -> grayscale:", colour, " binary:", colour_binary)
```

```python
gray_border = cv2.copyMakeBorder(gray, w, w, w, w, cv2.BORDER_CONSTANT, value=colour)
binary_border = cv2.copyMakeBorder(binary, w, w, w, w, cv2.BORDER_CONSTANT, value=colour_binary)

print("Before padding:", gray.shape)
print("After padding :", gray_border.shape)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1); plt.imshow(gray_border, cmap="gray");   plt.title("Grayscale + border"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(binary_border, cmap="gray"); plt.title("Binary + border");    plt.axis("off")
plt.show()
```

**2. WAP for Image Compliment.**

Try this for both binary and grayscale image.

$$Complement = 255 - \text{pixel value}$$

```python
gray_comp = 255 - gray
binary_comp = 255 - binary

plt.figure(figsize=(8, 8))
plt.subplot(2, 2, 1); plt.imshow(gray, cmap="gray");        plt.title("Grayscale");            plt.axis("off")
plt.subplot(2, 2, 2); plt.imshow(gray_comp, cmap="gray");   plt.title("Grayscale complement"); plt.axis("off")
plt.subplot(2, 2, 3); plt.imshow(binary, cmap="gray");      plt.title("Binary");               plt.axis("off")
plt.subplot(2, 2, 4); plt.imshow(binary_comp, cmap="gray"); plt.title("Binary complement");    plt.axis("off")
plt.show()
```
